```python
# ============================================================
# 01 - DATA UNDERSTANDING AND PROFILING
# PaySim Financial Transaction Fraud Analytics
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import gdown

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")

print("Libraries imported successfully.")


# ============================================================
# 2. LOAD RAW PAYSIM DATASET
# ============================================================

file_id = "1mxCeHEK7U1zj4mbhst1gV3KgrO7WHgpR"
file_path = "paysim.csv"

# Download only if the file does not already exist locally
if not os.path.exists(file_path):

    print("\nDataset not found locally.")
    print("Downloading PaySim dataset from Google Drive...")

    gdown.download(
        id=file_id,
        output=file_path,
        quiet=False
    )

else:
    print("\nDataset already exists locally.")

# Load dataset
df = pd.read_csv(file_path)

print("\nDataset loaded successfully.")
print(f"Dataset Shape: {df.shape}")


# ============================================================
# 3. DATASET DIMENSIONS
# ============================================================

print("\n" + "=" * 60)
print("3. DATASET DIMENSIONS")
print("=" * 60)

print(f"Number of Transactions: {df.shape[0]:,}")
print(f"Number of Variables: {df.shape[1]}")
print(f"Dataset Shape: {df.shape}")


# ============================================================
# 4. COLUMN NAMES
# ============================================================

print("\n" + "=" * 60)
print("4. COLUMN NAMES")
print("=" * 60)

for number, column in enumerate(df.columns, start=1):
    print(f"{number}. {column}")


# ============================================================
# 5. DATA TYPES
# ============================================================

print("\n" + "=" * 60)
print("5. DATA TYPES")
print("=" * 60)

dtype_summary = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str).values
})

display(dtype_summary)

print("\nDetailed Dataset Information:")
df.info()


# ============================================================
# 6. SAMPLE RECORDS
# ============================================================

print("\n" + "=" * 60)
print("6. SAMPLE RECORDS")
print("=" * 60)

print("\nFirst 5 Transactions:")
display(df.head())

print("\nRandom 5 Transactions:")
display(
    df.sample(
        n=5,
        random_state=42
    )
)


# ============================================================
# 7. DESCRIPTIVE STATISTICS
# ============================================================

print("\n" + "=" * 60)
print("7. DESCRIPTIVE STATISTICS")
print("=" * 60)

descriptive_stats = df.describe().T

display(
    descriptive_stats.style.format({
        "count": "{:,.0f}",
        "mean": "{:,.2f}",
        "std": "{:,.2f}",
        "min": "{:,.2f}",
        "25%": "{:,.2f}",
        "50%": "{:,.2f}",
        "75%": "{:,.2f}",
        "max": "{:,.2f}"
    })
)


# ============================================================
# 7.1 FINANCIAL VARIABLES DESCRIPTIVE STATISTICS
# ============================================================

print("\nFinancial Variables Descriptive Statistics")

financial_columns = [
    "amount",
    "oldbalanceOrg",
    "newbalanceOrig",
    "oldbalanceDest",
    "newbalanceDest"
]

financial_stats = df[financial_columns].describe().T

display(
    financial_stats.style.format({
        "count": "{:,.0f}",
        "mean": "{:,.2f}",
        "std": "{:,.2f}",
        "min": "{:,.2f}",
        "25%": "{:,.2f}",
        "50%": "{:,.2f}",
        "75%": "{:,.2f}",
        "max": "{:,.2f}"
    })
)


# ============================================================
# 8. MISSING VALUE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("8. MISSING VALUE ANALYSIS")
print("=" * 60)

missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage (%)": (
        df.isnull().mean() * 100
    ).round(4)
})

display(missing_summary)

total_missing = df.isnull().sum().sum()
columns_with_missing = df.isnull().any().sum()

print(f"Total Missing Values: {total_missing:,}")
print(f"Columns with Missing Values: {columns_with_missing}")

if total_missing == 0:
    print("Finding: No missing values were identified.")
else:
    print("Finding: Missing values require further investigation.")


# ============================================================
# 9. DUPLICATE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("9. DUPLICATE ANALYSIS")
print("=" * 60)

duplicate_count = df.duplicated().sum()

duplicate_percentage = (
    duplicate_count / len(df)
) * 100

print(f"Total Transactions: {len(df):,}")
print(f"Unique Transactions: {len(df.drop_duplicates()):,}")
print(f"Duplicate Transactions: {duplicate_count:,}")
print(f"Duplicate Percentage: {duplicate_percentage:.4f}%")

if duplicate_count > 0:

    print("\nSample Duplicate Records:")

    duplicate_records = df[
        df.duplicated(keep=False)
    ]

    display(duplicate_records.head(20))

else:
    print("Finding: No exact duplicate records were identified.")


# ============================================================
# 10. UNIQUE VALUES / CARDINALITY
# ============================================================

print("\n" + "=" * 60)
print("10. UNIQUE VALUES / CARDINALITY")
print("=" * 60)

cardinality = pd.DataFrame({
    "Column": df.columns,
    "Unique Values": df.nunique().values
})

cardinality["Unique Percentage (%)"] = (
    cardinality["Unique Values"]
    / len(df)
    * 100
).round(4)

display(cardinality)


# ============================================================
# 10.1 LOW-CARDINALITY VARIABLES
# ============================================================

print("\nLow-Cardinality Variables")
print("-" * 60)

for column in df.columns:

    unique_count = df[column].nunique()

    if unique_count <= 20:

        print(f"\nColumn: {column}")
        print(f"Unique Values: {unique_count}")

        print(
            df[column]
            .value_counts(dropna=False)
        )


# ============================================================
# 11. TRANSACTION TYPE DISTRIBUTION
# ============================================================

print("\n" + "=" * 60)
print("11. TRANSACTION TYPE DISTRIBUTION")
print("=" * 60)

transaction_distribution = pd.DataFrame({
    "Transaction Count":
        df["type"].value_counts(),

    "Percentage (%)":
        df["type"]
        .value_counts(normalize=True)
        .mul(100)
        .round(2)
})

display(transaction_distribution)


# Transaction type visualisation
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="type",
    order=df["type"].value_counts().index
)

plt.title(
    "Distribution of Transaction Types",
    fontsize=14
)

plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================================
# 12. FRAUD CLASS DISTRIBUTION
# ============================================================

print("\n" + "=" * 60)
print("12. FRAUD CLASS DISTRIBUTION")
print("=" * 60)

fraud_distribution = pd.DataFrame({
    "Transaction Count":
        df["isFraud"]
        .value_counts()
        .sort_index(),

    "Percentage (%)":
        df["isFraud"]
        .value_counts(normalize=True)
        .sort_index()
        .mul(100)
})

fraud_distribution.index = [
    "Legitimate",
    "Fraud"
]

fraud_distribution["Percentage (%)"] = (
    fraud_distribution["Percentage (%)"]
    .round(4)
)

display(fraud_distribution)


# Calculate class imbalance ratio
legitimate_count = (
    df["isFraud"] == 0
).sum()

fraud_count = (
    df["isFraud"] == 1
).sum()

imbalance_ratio = (
    legitimate_count / fraud_count
)

print(f"Legitimate Transactions: {legitimate_count:,}")
print(f"Fraudulent Transactions: {fraud_count:,}")
print(f"Imbalance Ratio: {imbalance_ratio:.2f}:1")


# Fraud class visualisation
plt.figure(figsize=(7, 5))

ax = sns.countplot(
    data=df,
    x="isFraud"
)

ax.set_xticks([0, 1])

ax.set_xticklabels([
    "Legitimate",
    "Fraud"
])

plt.title(
    "Fraud Class Distribution",
    fontsize=14
)

plt.xlabel("Transaction Class")
plt.ylabel("Number of Transactions")

plt.tight_layout()
plt.show()


# ============================================================
# 13. BUSINESS-RULE / LOGICAL VALIDATION
# ============================================================

print("\n" + "=" * 60)
print("13. BUSINESS-RULE / LOGICAL VALIDATION")
print("=" * 60)


# ============================================================
# 13.1 TRANSACTION AMOUNT VALIDATION
# ============================================================

print("\n13.1 Transaction Amount Validation")
print("-" * 60)

negative_amounts = (
    df["amount"] < 0
).sum()

zero_amounts = (
    df["amount"] == 0
).sum()

positive_amounts = (
    df["amount"] > 0
).sum()

print(f"Negative Amounts: {negative_amounts:,}")
print(f"Zero Amounts: {zero_amounts:,}")
print(f"Positive Amounts: {positive_amounts:,}")

print(
    f"Minimum Amount: "
    f"{df['amount'].min():,.2f}"
)

print(
    f"Maximum Amount: "
    f"{df['amount'].max():,.2f}"
)


# ============================================================
# 13.2 NEGATIVE BALANCE VALIDATION
# ============================================================

print("\n13.2 Negative Balance Validation")
print("-" * 60)

balance_columns = [
    "oldbalanceOrg",
    "newbalanceOrig",
    "oldbalanceDest",
    "newbalanceDest"
]

for column in balance_columns:

    negative_count = (
        df[column] < 0
    ).sum()

    print(
        f"{column}: "
        f"{negative_count:,} negative values"
    )


# ============================================================
# 13.3 TRANSACTION TYPE VALIDATION
# ============================================================

print("\n13.3 Transaction Type Validation")
print("-" * 60)

valid_types = [
    "CASH_IN",
    "CASH_OUT",
    "DEBIT",
    "PAYMENT",
    "TRANSFER"
]

invalid_transaction_types = (
    ~df["type"].isin(valid_types)
).sum()

print(
    "Observed Transaction Types:",
    sorted(df["type"].unique())
)

print(
    f"Invalid Transaction Types: "
    f"{invalid_transaction_types:,}"
)


# ============================================================
# 13.4 BINARY FRAUD INDICATOR VALIDATION
# ============================================================

print("\n13.4 Binary Fraud Indicator Validation")
print("-" * 60)

binary_columns = [
    "isFraud",
    "isFlaggedFraud"
]

for column in binary_columns:

    unique_values = sorted(
        df[column].unique()
    )

    invalid_count = (
        ~df[column].isin([0, 1])
    ).sum()

    print(f"\n{column}")
    print(f"Unique Values: {unique_values}")
    print(f"Invalid Values: {invalid_count:,}")


# ============================================================
# 13.5 ACCOUNT IDENTIFIER VALIDATION
# ============================================================

print("\n13.5 Account Identifier Validation")
print("-" * 60)

blank_origin = (
    df["nameOrig"]
    .astype(str)
    .str.strip()
    .eq("")
    .sum()
)

blank_destination = (
    df["nameDest"]
    .astype(str)
    .str.strip()
    .eq("")
    .sum()
)

same_account_count = (
    df["nameOrig"]
    == df["nameDest"]
).sum()

print(f"Blank Origin Accounts: {blank_origin:,}")
print(f"Blank Destination Accounts: {blank_destination:,}")

print(
    f"Same Origin/Destination Accounts: "
    f"{same_account_count:,}"
)


# ============================================================
# 14. BALANCE CONSISTENCY ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("14. BALANCE CONSISTENCY ANALYSIS")
print("=" * 60)

# Floating-point comparison tolerance
tolerance = 0.01


# ============================================================
# 14.1 OUTGOING TRANSACTION BALANCE CONSISTENCY
# ============================================================

outgoing_types = [
    "CASH_OUT",
    "TRANSFER",
    "PAYMENT",
    "DEBIT"
]

outgoing_mask = (
    df["type"].isin(outgoing_types)
)

outgoing_balance_error = (

    df.loc[
        outgoing_mask,
        "oldbalanceOrg"
    ]

    - df.loc[
        outgoing_mask,
        "amount"
    ]

    - df.loc[
        outgoing_mask,
        "newbalanceOrig"
    ]

).abs()

outgoing_inconsistencies = (
    outgoing_balance_error
    > tolerance
).sum()

print(
    "Outgoing Balance Inconsistencies:",
    f"{outgoing_inconsistencies:,}"
)

print("\nOutgoing Balance Error Statistics:")

display(
    outgoing_balance_error.describe()
)


# ============================================================
# 14.2 CASH-IN BALANCE CONSISTENCY
# ============================================================

cash_in_mask = (
    df["type"] == "CASH_IN"
)

cash_in_balance_error = (

    df.loc[
        cash_in_mask,
        "oldbalanceOrg"
    ]

    + df.loc[
        cash_in_mask,
        "amount"
    ]

    - df.loc[
        cash_in_mask,
        "newbalanceOrig"
    ]

).abs()

cash_in_inconsistencies = (
    cash_in_balance_error
    > tolerance
).sum()

print(
    "\nCASH_IN Balance Inconsistencies:",
    f"{cash_in_inconsistencies:,}"
)

print("\nCASH_IN Balance Error Statistics:")

display(
    cash_in_balance_error.describe()
)


# ============================================================
# 15. ZERO-VALUE TRANSACTION INVESTIGATION
# ============================================================

print("\n" + "=" * 60)
print("15. ZERO-VALUE TRANSACTION INVESTIGATION")
print("=" * 60)

zero_transactions = df[
    df["amount"] == 0
].copy()

print(
    f"Number of Zero-Value Transactions: "
    f"{len(zero_transactions):,}"
)

if len(zero_transactions) > 0:

    display(
        zero_transactions[
            [
                "step",
                "type",
                "amount",
                "nameOrig",
                "oldbalanceOrg",
                "newbalanceOrig",
                "nameDest",
                "oldbalanceDest",
                "newbalanceDest",
                "isFraud",
                "isFlaggedFraud"
            ]
        ]
    )

    print(
        "\nZero-Value Transactions "
        "by Transaction Type and Fraud Status:"
    )

    zero_transaction_summary = (
        zero_transactions
        .groupby(
            ["type", "isFraud"]
        )
        .size()
        .reset_index(
            name="Transaction Count"
        )
    )

    display(zero_transaction_summary)

else:

    print(
        "No zero-value transactions "
        "were identified."
    )


# ============================================================
# 16. BUSINESS-RULE VALIDATION SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("16. BUSINESS-RULE VALIDATION SUMMARY")
print("=" * 60)

invalid_isfraud = (
    ~df["isFraud"].isin([0, 1])
).sum()

invalid_flagged_fraud = (
    ~df["isFlaggedFraud"].isin([0, 1])
).sum()

negative_balance_count = (
    (df[balance_columns] < 0)
    .sum()
    .sum()
)

print(
    f"Negative Transaction Amounts: "
    f"{negative_amounts:,}"
)

print(
    f"Zero-Value Transactions: "
    f"{zero_amounts:,}"
)

print(
    f"Negative Balance Values: "
    f"{negative_balance_count:,}"
)

print(
    f"Invalid Transaction Types: "
    f"{invalid_transaction_types:,}"
)

print(
    f"Invalid isFraud Values: "
    f"{invalid_isfraud:,}"
)

print(
    f"Invalid isFlaggedFraud Values: "
    f"{invalid_flagged_fraud:,}"
)

print(
    f"Blank Origin Accounts: "
    f"{blank_origin:,}"
)

print(
    f"Blank Destination Accounts: "
    f"{blank_destination:,}"
)

print(
    f"Same Origin/Destination Accounts: "
    f"{same_account_count:,}"
)

print(
    f"Outgoing Balance Inconsistencies: "
    f"{outgoing_inconsistencies:,}"
)

print(
    f"CASH_IN Balance Inconsistencies: "
    f"{cash_in_inconsistencies:,}"
)


# ============================================================
# 17. DATA QUALITY SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("17. DATA QUALITY SUMMARY")
print("=" * 60)

data_quality_summary = pd.DataFrame({

    "Data Quality Check": [

        "Total Transactions",
        "Total Variables",
        "Missing Values",
        "Duplicate Transactions",

        "Negative Transaction Amounts",
        "Zero-Value Transactions",
        "Negative Balance Values",

        "Invalid Transaction Types",
        "Invalid isFraud Values",
        "Invalid isFlaggedFraud Values",

        "Blank Origin Accounts",
        "Blank Destination Accounts",
        "Same Origin/Destination Accounts",

        "Outgoing Balance Inconsistencies",
        "CASH_IN Balance Inconsistencies",

        "Maximum Transaction Amount"
    ],

    "Result": [

        f"{len(df):,}",
        f"{df.shape[1]:,}",
        f"{total_missing:,}",
        f"{duplicate_count:,}",

        f"{negative_amounts:,}",
        f"{zero_amounts:,}",
        f"{negative_balance_count:,}",

        f"{invalid_transaction_types:,}",
        f"{invalid_isfraud:,}",
        f"{invalid_flagged_fraud:,}",

        f"{blank_origin:,}",
        f"{blank_destination:,}",
        f"{same_account_count:,}",

        f"{outgoing_inconsistencies:,}",
        f"{cash_in_inconsistencies:,}",

        f"{df['amount'].max():,.2f}"
    ]
})

display(data_quality_summary)


# ============================================================
# 18. FINAL PROFILING SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("18. FINAL PROFILING SUMMARY")
print("=" * 60)

print(f"""
PaySim Dataset Profiling Completed

Dataset:
- Transactions: {len(df):,}
- Variables: {df.shape[1]}

Data Quality:
- Missing Values: {total_missing:,}
- Duplicate Transactions: {duplicate_count:,}
- Negative Transaction Amounts: {negative_amounts:,}
- Zero-Value Transactions: {zero_amounts:,}
- Negative Balance Values: {negative_balance_count:,}
- Invalid Transaction Types: {invalid_transaction_types:,}
- Invalid Fraud Labels: {invalid_isfraud:,}
- Blank Origin Accounts: {blank_origin:,}
- Blank Destination Accounts: {blank_destination:,}
- Same Origin/Destination Accounts: {same_account_count:,}

Fraud:
- Legitimate Transactions: {legitimate_count:,}
- Fraudulent Transactions: {fraud_count:,}
- Class Imbalance Ratio: {imbalance_ratio:.2f}:1

Further Investigation:
- Zero-value transactions
- Extreme transaction amounts
- Origin balance inconsistencies
- CASH_IN balance inconsistencies
- High-cardinality account identifiers
- Severe fraud class imbalance

No broad deletion, missing-value imputation, or duplicate
removal is justified based on the initial profiling results.
""")
```
